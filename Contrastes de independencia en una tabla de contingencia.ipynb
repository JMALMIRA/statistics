# Contrastes de Independencia en Tablas de Contingencia

A diferencia de la bondad de ajuste (que analiza una sola variable respecto a un modelo teórico), el **contraste de independencia** analiza dos variables cualitativas agrupadas en una tabla de doble entrada ($r$ filas y $c$ columnas).

## Fundamento Teórico

Sean $X$ e $Y$ dos variables categóricas. Se desea contrastar:

* $H_0$: $X$ e $Y$ son independientes ($P(X = i, Y = j) = P(X = i) \cdot P(Y = j)$).
* $H_1$: $X$ e $Y$ no son independientes (existe asociación entre ellas).

Bajo $H_0$, la frecuencia esperada en la celda $(i, j)$ se calcula a partir de los totales marginales:

$$E_{ij} = \frac{(\text{Total Fila } i) \times (\text{Total Columna } j)}{N}$$

El estadístico de prueba es idéntico en forma al de Pearson:

$$T = \sum_{i=1}^r \sum_{j=1}^c \frac{(O_{ij} - E_{ij})^2}{E_{ij}} \sim \chi^2_{(r-1)(c-1)}$$

## Regla de Cochran y Agrupación de Celdas

Al igual que en bondad de ajuste, la aproximación continua a la distribución $\chi^2$ requiere que las frecuencias esperadas sean suficientemente grandes. Se exige que $E_{ij} \ge 5$ (o al menos el $80\%$ de las celdas con $E_{ij} \ge 5$ y ninguna con $E_{ij} < 1$). Si no se cumple, **se deben agrupar filas o columnas adyacentes**.

# Ejemplo Práctico Completo

### Enunciado
Millie desea investigar si los estudiantes que cursaron distintas carreras científicas en la universidad reciben la misma remuneración al obtener un empleo. Para ello, realiza una encuesta a graduados en ciencias que se graduaron de la universidad en los últimos 5 años y registra la información relativa a sus salarios. Los resultados se recogen en la siguiente tabla:

| Ciencia estudiada | €0--€20k | €20k--€40k | €40k--€60k | €60k--€80k | >€80k | Totales |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: |
| **Biología** | 4 | 69 | 23 | 5 | 3 | **104** |
| **Química** | 3 | 72 | 27 | 4 | 2 | **108** |
| **Física** | 2 | 68 | 32 | 5 | 4 | **111** |
| **Totales** | **9** | **209** | **82** | **14** | **9** | **323** |

Ella realiza un contraste al nivel de significación del 5% para determinar si existe asociación entre la ciencia estudiada y el salario.

* **a)** Indique la hipótesis nula y la hipótesis alternativa para este contraste.
* **b)** Calcule el estadístico de prueba para este contraste y determine, justificando la respuesta, si se rechaza o no la hipótesis nula.

### Paso 1: Planteamiento de Hipótesis
* $H_0$: No existe asociación entre la carrera científica estudiada y el salario obtenido.
* $H_1$: Existe asociación entre la carrera científica estudiada y el salario obtenido.

### Paso 2: Comprobación de $E_{ij}$ y Agrupación de Columnas
Calculando las frecuencias esperadas para la tabla original $3 \times 5$, se detecta que las columnas de salarios extremos («€0--€20k», «€60k--€80k» y «>€80k») presentan valores de $E_{ij} < 5$ (p. ej., $E_{\text{Biol, 0-20k}} = 2.90 < 5$).

Siguiendo la regla de Cochran, agrupamos:
* «€0--€20k» + «€20k--€40k» $\longrightarrow$ **< €40k**
* «€60k--€80k» + «>€80k» $\longrightarrow$ **> €60k**

| Carrera | < €40k | €40k--€60k | > €60k | Total |
| :--- | :---: | :---: | :---: | :---: |
| **Biología** | $73$ ($70.19$) | $23$ ($26.40$) | $8$ ($7.41$) | **104** |
| **Química** | $75$ ($72.89$) | $27$ ($27.42$) | $6$ ($7.69$) | **108** |
| **Física** | $70$ ($74.92$) | $32$ ($28.18$) | $9$ ($7.90$) | **111** |
| **Total** | **218** | **82** | **23** | **323** |

### Paso 3: Cálculo del Estadístico $T$
Sumando las 9 celdas reestructuradas:

$$T = \sum_{i=1}^3 \sum_{j=1}^3 \frac{(O_{ij} - E_{ij})^2}{E_{ij}} = 0.113 + 0.438 + 0.047 + 0.061 + 0.007 + 0.372 + 0.323 + 0.518 + 0.153 = 2.032$$

### Paso 4: Región Crítica y Decisión
* **Grados de libertad:** $\text{gl} = (r - 1)(c - 1) = (3 - 1)(3 - 1) = 4$.
* **Valor crítico ($\alpha = 0.05$):** $\chi^2_{0.05, 4} = 9.488$.

**Conclusión:** Como $T = 2.032 < 9.488$, **no rechazamos la hipótesis nula $H_0$**. No existe evidencia estadística suficiente para afirmar que el salario depende de la carrera científica cursada.

# Implementación de la Prueba de Independencia en Python

Para contrastes de independencia se utiliza la función `scipy.stats.chi2_contingency`. Esta función calcula automáticamente las frecuencias esperadas $E_{ij}$ y los grados de libertad a partir de la matriz de frecuencias observadas.

```python
import numpy as np
from scipy import stats

# 1. Matriz de frecuencias observadas tras agrupar columnas (3x3)
# Filas: Biologia, Quimica, Fisica
# Columnas: <40k, 40k-60k, >60k
O_grouped = np.array([
    [73, 23, 8],  # Biologia
    [75, 27, 6],  # Quimica
    [70, 32, 9]   # Fisica
])

# 2. Ejecucion del contraste de independencia Chi-cuadrado
# correction=False evita la correccion de continuidad de Yates (usada en 2x2)
stat, p_value, dof, expected = stats.chi2_contingency(O_grouped, correction=False)

print("--- RESULTADOS CONTRASTE DE INDEPENDENCIA ---")
print(f"Estadistico Chi2 calculado: {stat:.3f}")
print(f"Grados de libertad: {dof}")
print(f"p-valor: {p_value:.4f}")

print("\nMatriz de Frecuencias Esperadas (E_ij):")
print(np.round(expected, 2))

# 3. Decision estadistica
alpha = 0.05
crit_val = stats.chi2.ppf(1 - alpha, dof)
print(f"\nValor critico para alpha=0.05: {crit_val:.3f}")

if p_value > alpha:
    print("Conclusion: No se rechaza H0. No hay asociacion significativa entre carrera y salario.")
else:
    print("Conclusion: Se rechaza H0. Existe asociacion significativa entre carrera y salario.")
```
