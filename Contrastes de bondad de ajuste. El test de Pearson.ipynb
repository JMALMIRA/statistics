# Motivación del Estadístico de Pearson

El objetivo de un contraste de bondad de ajuste es determinar si una muestra de frecuencias observadas $O_1, O_2, \dots, O_m$ proviene de un modelo teórico con probabilidades fijadas $p_1, p_2, \dots, p_m$. Las frecuencias esperadas teóricas bajo la hipótesis nula ($H_0$) son $E_i = n \cdot p_i$, con $n = \sum_{i=1}^m O_i$.

Para medir la discrepancia global, Karl Pearson propuso el estadístico:

$$T = \sum_{i=1}^m \frac{(O_i - E_i)^2}{E_i}$$

La estructura del estadístico responde a tres principios fundamentales:

1. **Diferencia simple $(O_i - E_i)$:** Mide el error directo. No se puede sumar directamente porque las desviaciones se anulan mutuamente ($\sum (O_i - E_i) = n - n = 0$).
2. **Elevar al cuadrado $(O_i - E_i)^2$:** Elimina signos negativos y penaliza con mayor severidad las desviaciones grandes.
3. **División por $E_i$ (Estandarización):** Escala el error cuadrático respecto al tamaño esperado de la categoría. Una desviación de 10 unidades es crítica si se esperaban 12 casos ($E_i=12$), pero irrelevante si se esperaban 10 000 ($E_i=10 000$).

# Relación con la Distribución $\chi^2$ y Restricción $E_i \ge 5$

Si asumimos que las observaciones provienen del modelo teórico propuesto, entonces la distribución de cada frecuencia observada $O_i$ sigue una binomial, $O_i \sim Bin(n, p_i)$. Por tanto, esta se aproxima a una Normal cuando $n$ es grande (esto puede verse directamente, pero también como consecuencia del Teorema Central del Límite, pues una binomial es una suma de v.a. de tipo Bernoulli independientes). En particular:

$$O_i \sim \mathcal{N}(n p_i, n p_i(1 - p_i)) = \mathcal{N}(E_i, E_i(1 - p_i))$$

y, por tanto, 

$$\frac{O_i - E_i}{\sqrt{E_i(1 - p_i)}} \sim Z = \mathcal{N}(0, 1)$$

Veamos qué sucede con el estadístico de Pearson cuando tomamos exactamente $m = 2$ celdas, que es el caso más sencillo posible. En tal caso:

$$T = \frac{(O_1 - E_1)^2}{E_1} + \frac{(O_2 - E_2)^2}{E_2}$$

Ahora bien, la desviación en la celda de fracasos es la opuesta a la de éxitos:

$$O_2 - E_2 = (n - O_1) - n(1 - p_1) = -(O_1 - n p_1) = -(O_1 - E_1)$$

Por tanto, al elevar al cuadrado, ambos numeradores coinciden:

$$(O_2 - E_2)^2 = (O_1 - E_1)^2$$

y podemos reescribir el estadístico de Pearson como:

$$T = (O_1 - E_1)^2 \left( \frac{1}{E_1} + \frac{1}{E_2} \right)$$

Desarrollando los denominadores:

$$T = (O_1 - E_1)^2 \left( \frac{1}{n p_1} + \frac{1}{n(1 - p_1)} \right) = (O_1 - E_1)^2 \left( \frac{(1 - p_1) + p_1}{n p_1 (1 - p_1)} \right)$$

$$T = \frac{(O_1 - E_1)^2}{n p_1 (1 - p_1)} = \frac{(O_1 - E_1)^2}{\text{Var}(O_1)} = Z^2 \sim \chi^2_1$$

Este tipo de cálculos se puede generalizar al caso de $m \ge 3$ celdas (pero no es trivial), de modo que $T$ se puede expresar (si la única restricción es $\sum_{i=1}^m O_i = n$) como una <font color='orange'>suma de $m-1$ cuadrados de normales independientes</font>, por lo que:

$$T \sim \chi^2_{m-1}$$

(decimos que $m-1$ celdas son libres o que disponemos de $m-1$ grados de libertad). Si además estimamos $p$ parámetros del modelo a partir de los datos, entonces solo $m-1-p$ celdas son libres, por lo que tendremos:

$$T \sim \chi^2_{m-1-p}$$

---

## La Regla de Cochran ($E_i \ge 5$)

Dado que la aproximación a la Normal depende del tamaño muestral en cada celda, si $E_i < 5$:
- La distribución de $O_i$ conserva su discreción y asimetría original.
- La aproximación continua a la distribución $\chi^2$ se distorsiona.

**Regla práctica (Cochran):** Ninguna celda debe tener $E_i < 1$ y al menos el $80\%$ de las celdas deben satisfacer $E_i \ge 5$. En la práctica académica estricta, se exige $E_i \ge 5$ para todas las celdas.

Si $E_i < 5$, se deben **agrupar o combinar celdas contiguas o afines**. Al combinar las celdas $j$ y $k$, las nuevas frecuencias son:

$$O_{\text{nueva}} = O_j + O_k, \quad E_{\text{nueva}} = E_j + E_k$$

**Consecuencia directa:** Si el número inicial de celdas $m$ se reduce a $m'$ categorías tras la agrupación, los grados de libertad cambian automáticamente a $\text{gl} = m' - 1 - p$.

# Ejemplo Práctico Completo con Agrupación de Celdas

Se analiza el número de accidentes laborales por día durante un mes en una planta industrial ($n = 100$ días) para verificar si siguen una distribución de Poisson con media $\lambda = 1.5$.

Las probabilidades teóricas bajo $H_0: X \sim \text{Poisson}(\lambda = 1.5)$ son $p_x = \frac{e^{-1.5} 1.5^x}{x!}$.

## Paso 1: Cálculo de Frecuencias Esperadas Iniciales

| Accidentes ($x$) | $O_x$ | Probabilidad $p_x$ | $E_x $ | Condición $E_x \ge 5$ |
| :---: | :---: | :---: | :---: | :---: |
| 0 | 25 | 0.2231 | 22.31 | Sí ($\ge 5$) |
| 1 | 38 | 0.3347 | 33.47 | Sí ($\ge 5$) |
| 2 | 22 | 0.2510 | 25.10 | Sí ($\ge 5$) |
| 3 | 10 | 0.1255 | 12.55 | Sí ($\ge 5$) |
| 4 | 3 | 0.0471 | 4.71 | **No ($< 5$)** |
| $\ge 5$ | 2 | 0.0186 | 1.86 | **No ($< 5$)** |
| **Total** | **100** | **1.0000** | **100.00** | $m = 6$ celdas |

## Paso 2: Agrupación de Celdas con $E_x < 5$

Las celdas $x = 4$ ($E_4 = 4.71$) y $x \ge 5$ ($E_{\ge 5} = 1.86$) no cumplen el criterio $E_x \ge 5$. Las combinamos en una sola categoría: **"$x \ge 4$"**.

- $O_{\ge 4} = O_4 + O_{\ge 5} = 3 + 2 = 5$
- $E_{\ge 4} = E_4 + E_{\ge 5} = 4.71 + 1.86 = 6.57 \quad (\ge 5)$

El número final de categorías tras agrupar se reduce de $m = 6$ a $m' = 5$.

## Paso 3: Cálculo del Estadístico $T$ con Celdas Reestructuradas

| Categoría ($i$) | Frec. observada: $O_i$ |Frec. esperada: $E_i$ | Diferencia: $O_i-E_i$ | $\frac{(O_i - E_i)^2}{E_i}$ |
| :---: | :---: | :---: | :---: | :---: |
| 0 | 25 | 22.31 | 2.69 | $7.2361 / 22.31 = 0.324$ |
| 1 | 38 | 33.47 | 4.53 | $20.5209 / 33.47 = 0.613$ |
| 2 | 22 | 25.10 | -3.10 | $9.6100 / 25.10 = 0.383$ |
| 3 | 10 | 12.55 | -2.55 | $6.5025 / 12.55 = 0.518$ |
| $\ge 4$ | 5 | 6.57 | -1.57 | $2.4649 / 6.57 = 0.375$ |
| **Total** | **100** | **100.00** | **0.00** | **$T = 2.213$** |

El valor calculado del estadístico es **$T = 2.213$**.

## Paso 4: Región Crítica y Decisión

- **Categorías reducidas ($m'$):** 5
- **Parámetros estimados ($p$):** 0 (se asumió $\lambda = 1.5$ prefijado)
- **Grados de libertad:** $\text{gl} = m' - 1 - p = 5 - 1 - 0 = 4$
- **Valor crítico ($\alpha = 0.05$):** Buscando en la tabla $\chi^2$ con 4 gl:

$$\chi^2_{0.05, 4} = 9.488$$

**Conclusión:**
Como el estadístico calculado $T = 2.213$ es menor que el valor crítico $9.488$, **no rechazamos la hipótesis nula**. Se concluye con un $95\%$ de confianza que la distribución de accidentes se ajusta adecuadamente a una distribución de Poisson con $\lambda = 1.5$.

# Implementación en Python

Para realizar este contraste en Python se utiliza la función `scipy.stats.chisquare`. Al usar esta librería es crucial considerar dos aspectos:

1. **Agrupación previa:** La función `chisquare` **no** realiza el colapsado de celdas con $E_i < 5$ de forma automática. Debe realizarse manualmente sobre los vectores de frecuencias antes de ejecutar la función.
2. **Ajuste de Grados de Libertad (`ddof`):** El parámetro `ddof` (*Delta Degrees of Freedom*) resta grados de libertad adicionales. Por defecto es $0$ (calcula $\text{gl} = m' - 1$). Si se estiman $p$ parámetros a partir de la muestra, se debe indicar `ddof=p` para que compute $\text{gl} = m' - 1 - p$.

```python
import numpy as np
from scipy import stats
```

```python

# 1. Datos del experimento
n = 100
lambda_p = 1.5
O_orig = np.array([25, 38, 22, 10, 3, 2]) # Frecuencias observadas (0, 1, 2, 3, 4, >=5)
print("Frecuencias observadas: ", O_orig)
```

```python
# 2. Probabilidades teoricas Poisson(1.5)
# Para x = 0, 1, 2, 3, 4
p_orig = [stats.poisson.pmf(k, lambda_p) for k in range(5)]
# La ultima probabilidad incluye el resto de la cola: P(X >= 5) = 1 - P(X <= 4)
p_orig.append(1 - stats.poisson.cdf(4, lambda_p))
p_orig = np.array(p_orig)

# Frecuencias esperadas iniciales
E_orig = n * p_orig
print("Frecuencias esperadas: ", E_orig)
```

```python

# 3. Agrupacion manual de las celdas con E_i < 5 (indices 4 y 5)
O_grouped = np.array([O_orig[0], O_orig[1], O_orig[2], O_orig[3], O_orig[4] + O_orig[5]])
E_grouped = np.array([E_orig[0], E_orig[1], E_orig[2], E_orig[3], E_orig[4] + E_orig[5]])

print("Frecuencias observadas, agrupadas: ", O_grouped )
print("Frecuencias esperadas, agrupadas: ", E_grouped )
```

```python
# 4. Calculo del contraste Chi-cuadrado
# p = 0 porque lambda era fijo (no estimado de la muestra), por tanto ddof=0
p_estimados = 0 
stat, p_valor = stats.chisquare(f_obs=O_grouped, f_exp=E_grouped, ddof=p_estimados)
```

```python
# Grados de libertad resultantes y valor crítico
df = len(O_grouped) - 1 - p_estimados
crit_val = stats.chi2.ppf(0.95, df)
```

```python
print(f"Estadistico T calculado: ", stat)
print(f"Grados de libertad: ", df)
print(f"Valor critico (alpha=0.05):", crit_val)
print(f"p-valor: ", p_valor)
```

```python
if p_valor > 0.05:
    print("Conclusion: No se rechaza H0. Los datos siguen la distribucion Poisson(1.5).")
else:
    print("Conclusion: Se rechaza H0.")
```
